# Project Setup and Initialization

Learn how to set up standardized PyTheranostics projects with configuration templates and organized directory structures.

## Overview

PyTheranostics provides project initialization tools to help you:
- Create standardized project structures
- Get pre-configured template files
- Organize your data, results, and analysis outputs
- Start with best-practice workflows

This tutorial covers:
1. **Quick project initialization** - Get started in seconds
2. **Configuration templates** - Understanding available templates
3. **Customization options** - Tailoring projects to your needs
4. **Practical workflows** - Using initialized projects effectively

---

````python
# Import the project initialization tools
from pytheranostics import init_project, list_templates
from pytheranostics.project import get_template_path
import json
from pathlib import Path
````

## Step 1: List Available Templates

Before creating a project, let's see what configuration templates are available:

````python
# List all available configuration templates
templates = list_templates()

print("Available Configuration Templates:\n" + "="*50)
for name, description in templates.items():
    print(f"📄 {name}")
    print(f"   └─ {description}\n")
````

## Step 2: Inspect a Template

Let's look at what's inside the `total_seg_config.json` template:

````python
# Get the template path
template_path = get_template_path("total_seg_config.json")
print(f"Template location: {template_path}\n")

# Load and preview the first few entries
with open(template_path) as f:
    config = json.load(f)

print("Template structure:")
print(f"  - Number of VOIs defined: {len(config.get('vois', []))}")
print(f"  - Has 'combine' section: {'combine' in config}")

print("\nFirst 5 VOIs:")
for voi in config['vois'][:5]:
    included = "✓" if voi['include'] else "✗"
    name = voi.get('new_name') or voi['voi_name']
    print(f"  {included} {voi['voi_name']} → {name}")
````

## Step 3: Create Your First Project

Now let's initialize a complete project with all templates and standard directories:

````python
# Create a new project
project_path = init_project(
    "./my_dosimetry_study",
    project_name="Lu-177 Dosimetry Study",
    create_subdirs=True,
    overwrite=False  # Won't overwrite if already exists
)

print(f"\n✓ Project created at: {project_path}")
````

### Project Structure Created

The `init_project()` function created:

```
my_dosimetry_study/
├── total_seg_config.json       # TotalSegmentator configuration
├── voi_mappings_config.json    # VOI name mappings
├── README.md                    # Project documentation
├── data/                        # Raw DICOM data
├── results/                     # Analysis outputs
├── segmentations/              # TotalSegmentator outputs
├── rtstructs/                  # RT-STRUCT DICOM files
└── notebooks/                  # Jupyter notebooks
```

Let's verify what was created:

````python
# List the created files and directories
if project_path.exists():
    print(f"Contents of {project_path}:\n")
    
    print("Configuration Files:")
    for item in sorted(project_path.iterdir()):
        if item.is_file():
            size_kb = item.stat().st_size / 1024
            print(f"  📄 {item.name} ({size_kb:.1f} KB)")
    
    print("\nDirectories:")
    for item in sorted(project_path.iterdir()):
        if item.is_dir():
            print(f"  📁 {item.name}/")
````

## Step 4: Understanding Configuration Files

### TotalSegmentator Configuration

The `total_seg_config.json` controls which organs to include in RT-STRUCT files:

````python
# Load the project's TotalSegmentator config
config_path = project_path / "total_seg_config.json"

with open(config_path) as f:
    config = json.load(f)

# Show how many organs are included by default
included_vois = [v for v in config['vois'] if v['include']]
excluded_vois = [v for v in config['vois'] if not v['include']]

print(f"TotalSegmentator Configuration:")
print(f"  Total structures: {len(config['vois'])}")
print(f"  Included by default: {len(included_vois)}")
print(f"  Excluded by default: {len(excluded_vois)}")

print(f"\nIncluded structures (showing first 10):")
for voi in included_vois[:10]:
    name = voi.get('new_name') or voi['voi_name']
    print(f"  ✓ {voi['voi_name']} → {name}")
````

### VOI Mappings Configuration

The `voi_mappings_config.json` helps map organ names between different conventions:

````python
# Load the VOI mappings config
mappings_path = project_path / "voi_mappings_config.json"

with open(mappings_path) as f:
    mappings = json.load(f)

print("VOI Mappings Configuration:\n")

print("CT Mappings (morphology-based, suffix _m):")
for source, target in list(mappings['ct_mappings'].items())[:5]:
    print(f"  {source} → {target}")

print("\nSPECT Mappings (activity-based, suffix _a):")
for source, target in list(mappings['spect_mappings'].items())[:5]:
    print(f"  {source} → {target}")

print("\nThis helps unify naming across different data sources!")
````

## Step 6: Advanced - Selective Initialization

You can also create minimal projects with only specific templates:

````python
# Create a minimal project with only TotalSegmentator config
minimal_project = init_project(
    "./minimal_segmentation_project",
    templates=['total_seg_config.json'],  # Only this template
    create_subdirs=False,  # No standard directories
    overwrite=True
)

print(f"✓ Minimal project created at: {minimal_project}")
print("\nContents:")
for item in sorted(minimal_project.iterdir()):
    icon = "📄" if item.is_file() else "📁"
    print(f"  {icon} {item.name}")
````

## API Reference

### init_project()

```python
init_project(
    project_dir: str | Path,
    project_name: Optional[str] = None,
    templates: Optional[List[str]] = None,
    create_subdirs: bool = True,
    overwrite: bool = False,
) -> Path
```

**Parameters**:
- `project_dir`: Path where project should be created
- `project_name`: Project name (defaults to directory name)
- `templates`: List of template names to copy (defaults to all)
- `create_subdirs`: Create standard directories (default: True)
- `overwrite`: Overwrite existing config files (default: False)

**Returns**: Path to created project directory

### list_templates()

```python
list_templates() -> dict
```

**Returns**: Dictionary mapping template names to descriptions

### get_template_path()

```python
get_template_path(template_name: str) -> Path
```

**Parameters**:
- `template_name`: Name of template file

**Returns**: Path to template file within package

**Raises**: FileNotFoundError if template doesn't exist
